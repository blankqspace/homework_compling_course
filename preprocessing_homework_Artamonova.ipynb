```python
# pymorphy
import pymorphy3
from pymystem3 import Mystem

no_links = (re.sub(r'\[\d+\]', '', text)).strip()
mystem = Mystem(disambiguation=True)
morph = pymorphy3.MorphAnalyzer()

tokens = []
for token in mystem.analyze(no_links):
    if token.get('text'): # and token['text'].strip():
       tokens.append(token['text'])

lemma_pymorphy = []
for token in tokens:
  parsed = morph.parse(token)[0]
  norm = parsed.normal_form
  lemma_pymorphy.append(norm)

print(lemma_pymorphy)
```
